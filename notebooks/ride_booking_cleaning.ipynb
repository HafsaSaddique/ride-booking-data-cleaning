```python
cleaned_df = df.copy()

print("Original rows:", len(df))
print("Cleaning dataframe rows:", len(cleaned_df))
```

```python
cleaned_df["Date"] = pd.to_datetime(cleaned_df["Date"])

print(cleaned_df["Date"].dtype)
```

```python
cleaned_df = cleaned_df.drop(columns=["Unnamed: 20"])

print("Columns after removing empty column:", len(cleaned_df.columns))
```

```python
cleaned_df = cleaned_df.drop(columns=["Vehicle Images"])

print("Columns after removing unusable column:", len(cleaned_df.columns))
```

```python
cleaned_df.shape
```

```python
cleaned_df.dtypes
```

```python
cleaned_df.isnull().sum()
```

```python
cleaned_df.head()
```

```python
print("Non-successful bookings:")
print((cleaned_df["Booking_Status"] != "Success").sum())

print("\nMissing V_TAT in non-successful bookings:")
print(
    cleaned_df.loc[
        cleaned_df["Booking_Status"] != "Success", "V_TAT"
    ].isna().sum()
)

print("\nMissing C_TAT in non-successful bookings:")
print(
    cleaned_df.loc[
        cleaned_df["Booking_Status"] != "Success", "C_TAT"
    ].isna().sum()
)

print("\nMissing Driver Ratings in non-successful bookings:")
print(
    cleaned_df.loc[
        cleaned_df["Booking_Status"] != "Success", "Driver_Ratings"
    ].isna().sum()
)

print("\nMissing Customer Ratings in non-successful bookings:")
print(
    cleaned_df.loc[
        cleaned_df["Booking_Status"] != "Success", "Customer_Rating"
    ].isna().sum()
)
```

```python
categorical_na_columns = [
    "Canceled_Rides_by_Customer",
    "Canceled_Rides_by_Driver",
    "Incomplete_Rides",
    "Incomplete_Rides_Reason",
    "Payment_Method"
]

for column in categorical_na_columns:
    cleaned_df[column] = cleaned_df[column].fillna("Not Applicable")
```

```python
cleaned_df[categorical_na_columns].isnull().sum()
```

```python
date_time_match = (
    cleaned_df["Date"].dt.strftime("%H:%M:%S")
    == cleaned_df["Time"]
)

print("Rows where Date time matches Time:", date_time_match.sum())
print("Total rows:", len(cleaned_df))
```

```python
category_columns = [
    "Booking_Status",
    "Vehicle_Type",
    "Pickup_Location",
    "Drop_Location",
    "Payment_Method"
]

for column in category_columns:
    print(f"\n--- {column} ---")
    print(cleaned_df[column].value_counts(dropna=False).head(30))
```

```python
cleaned_df = cleaned_df.drop(columns=["Time"])

print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
```

```python
text_columns = [
    "Booking_Status",
    "Vehicle_Type",
    "Pickup_Location",
    "Drop_Location",
    "Canceled_Rides_by_Customer",
    "Canceled_Rides_by_Driver",
    "Incomplete_Rides",
    "Incomplete_Rides_Reason",
    "Payment_Method"
]

for column in text_columns:
    whitespace_count = (
        cleaned_df[column].astype(str)
        .ne(cleaned_df[column].astype(str).str.strip())
        .sum()
    )

    print(f"{column}: {whitespace_count} values with extra spaces")
```

```python
print("Total rides with 0 distance:")
print((cleaned_df["Ride_Distance"] == 0).sum())

print("\nSuccessful rides with 0 distance:")
print(
    (
        (cleaned_df["Ride_Distance"] == 0) &
        (cleaned_df["Booking_Status"] == "Success")
    ).sum()
)

print("\nNon-successful rides with 0 distance:")
print(
    (
        (cleaned_df["Ride_Distance"] == 0) &
        (cleaned_df["Booking_Status"] != "Success")
    ).sum()
)
```

```python
print("V_TAT outside expected positive range:")
print(((cleaned_df["V_TAT"].notna()) & (cleaned_df["V_TAT"] <= 0)).sum())

print("\nC_TAT outside expected positive range:")
print(((cleaned_df["C_TAT"].notna()) & (cleaned_df["C_TAT"] <= 0)).sum())

print("\nBooking Value <= 0:")
print((cleaned_df["Booking_Value"] <= 0).sum())

print("\nRide Distance < 0:")
print((cleaned_df["Ride_Distance"] < 0).sum())

print("\nDriver Rating outside 3-5:")
print(
    (
        cleaned_df["Driver_Ratings"].notna() &
        ~cleaned_df["Driver_Ratings"].between(3, 5)
    ).sum()
)

print("\nCustomer Rating outside 3-5:")
print(
    (
        cleaned_df["Customer_Rating"].notna() &
        ~cleaned_df["Customer_Rating"].between(3, 5)
    ).sum()
)
```

```python
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])

print("\nColumn names:")
print(cleaned_df.columns.tolist())

print("\nMissing values:")
print(cleaned_df.isnull().sum())
```
